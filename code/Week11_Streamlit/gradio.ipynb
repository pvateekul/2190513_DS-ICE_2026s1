##**Standard demos:**

which have both separate inputs and outputs (e.g. an image classifier or speech-to-text model)

```python
import numpy as np
import gradio as gr

def sepia(input_img):
    sepia_filter = np.array([
        [0.393, 0.769, 0.189],
        [0.349, 0.686, 0.168],
        [0.272, 0.534, 0.131]
    ])
    sepia_img = input_img.dot(sepia_filter.T)
    sepia_img /= sepia_img.max()
    return sepia_img

demo = gr.Interface(sepia, gr.Image(), "image", api_name="predict")
demo.launch()
```

##**Using Hugging Face Integrations**

**translation**

```python
from transformers import pipeline
import gradio as gr

pipe = pipeline("translation", model="Helsinki-NLP/opus-mt-en-es")

demo = gr.Interface.from_pipeline(pipe)
demo.launch(share=True)
```

**text to speech**

```python
from transformers import pipeline
import gradio as gr

tts = pipeline("text-to-speech", model="facebook/mms-tts-eng")

def speak(text):
    out = tts(text)
    return out["sampling_rate"], out["audio"].squeeze()

demo = gr.Interface(fn=speak, inputs="text", outputs="audio",
                    title="Text to Speech", flagging_mode="never")
demo.launch(share=True, debug=True)
```
